```python
#Reverse number
a=int(input("Enter number- "))
rev=0;
while a>0:
  rem=a%10
  rev=rev*10+rem
  a=a//10
print(rev)
```

```python
#Palindrome check
a=int(input("Enter number- "))
og=a
rev=0;
while a>0:
  rem=a%10
  rev=rev*10+rem
  a=a//10
if og==rev:
  print("Palindrome")
else:
  print("Not Palindrome+")

```
