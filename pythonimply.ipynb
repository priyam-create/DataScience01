```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
```

```python
#Dataset: 100 Students
np.random.seed(42)
```

```python
df=pd.DataFrame({
'Student_ID' : range(1,101),
'Name' : [f'Student_{i}' for i in range (1,101)],
'Gender' : np.random.choice(['Male','Female'],100),
'Department' : np.random.choice(['CSE','ECE','ME','CE'],100),
'Marks' : np.random.randint(40, 101,100)
})
```

```python
df
```

```python
print("Population Size",len(df))
```

```python
#About dataset
print("Department Distribution",df['Department'].value_counts())
#Gender Distribution
print("\nGender Distribution",df['Gender'].value_counts())
#Marks Distribution
print("\nBasic Distribution of Marks ",df['Marks'].describe())
```

```python
sample_size=12

simple_random_sample=df.sample(
    n=sample_size,
    random_state=42
)
```

```python
simple_random_sample
```

```python
#Visualization for sample Random selection
plt.figure(figsize=(12,6))
plt.scatter(df['Student_ID'],df['Marks'],alpha=0.35,label='population',color='Red')
plt.scatter(simple_random_sample['Student_ID'],simple_random_sample['Marks'],s=120,label='Selected Sample')
plt.xlabel('Student ID')
plt.ylabel('Marks')
plt.title('Simple random sample')
plt.legend()
plt.show()
```
