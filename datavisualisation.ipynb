```python
import pandas as pd
```

```python
#line chart
df=pd.DataFrame({
            "Department":['CSE','ECE','IT','ME'],
            "Students":[50,30,20,10]
})
df.plot(x='Department',y='Students',kind='line',marker='o')
```

```python
#horizontal bar
df=pd.DataFrame({
            "Department":['CSE','ECE','IT','ME'],
            "Students":[50,30,20,10]
})
df.plot(x='Department',y='Students',kind='barh',color=['red','blue','green','yellow'])
```

```python
df=pd.DataFrame({
            "Department":['CSE','ECE','IT','ME'],
            "Students":[50,30,20,10]
})
df.plot(x='Department',y='Students',kind='pie')
```

```python
df=pd.DataFrame({
            "Department":['CSE','ECE','IT','ME'],
            "Students":[50,30,20,10]
})
df.plot(x='Department',y='Students',kind='hist')
```

```python
df=pd.DataFrame({
            "Department":['CSE','ECE','IT','ME'],
            "Students":[50,30,20,10]
})
df.plot(x='Department',y='Students',kind='box')
```

```python
df=pd.DataFrame({
            "Department":['CSE','ECE','IT','ME'],
            "Students":[50,30,20,10]
})
df.plot(x='Department',y='Students',kind='scatter',color='red')
```

```python
import matplotlib.pyplot as plt
```

```python
roko=pd.read_csv("/content/sharma-kohli.csv")
```

```python
roko
```

```python
from matplotlib.lines import lineStyles
from matplotlib import legend
plt.plot(roko['index'],roko['V Kohli'],label='King Kohli',color='red',marker='s',ls='-.',linewidth=2)
plt.plot(roko['index'],roko['RG Sharma'],label='Ro-HIT Sharma',color='blue',marker='o',ls=':',lw=2.5)
plt.legend(loc='lower center')
plt.title("Rohit Sharma vs Virat Kohli")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.ylim(100,1000)
plt.grid(True,linestyle=":",alpha=0.5)
plt.savefig("Rohit vs Virat.png")
plt.figure(figsize=(10,5))
plt.show()
```

```python

df1 = pd.DataFrame({
    "Over": range(1, 21),
    "Runs": [3, 10, 7, 11, 7, 11, 9, 4, 9, 4, 13, 2, 18, 13, 4, 6, 6, 12, 4, 14]
})

df2 = pd.DataFrame({
    "Over": range(1, 20),
    "Runs": [6, 5, 15, 9, 4, 9, 7, 7, 6, 9, 7, 11, 3, 1, 4, 11, 2, 2, 1]
})
```

```python
plt.plot(df1["Over"], df1["Runs"], label="India", color="blue", linewidth=2.5)
plt.plot(df2["Over"], df2["Runs"], label="Australia", color="gold", linewidth=2.5)

plt.xlabel("Over")
plt.ylabel("Runs")
plt.title("IND vs AUS: Runs Per Over")
plt.legend()
plt.grid(True, linewidth=0.5)
plt.show()
```
